# 🎛️ AI Stem Separator (Demucs)

Separate any song into **vocals, drums, bass and other** using Meta's [Demucs](https://github.com/facebookresearch/demucs) (`htdemucs` model).

> 💡 For faster processing: *Runtime → Change runtime type → GPU*.

### 2. Upload your audio file

```python
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]
print(f"Uploaded: {filename}")
```

### 3. Separate stems with Demucs

Set `TWO_STEMS = True` to get only *vocals* + *instrumental*.

```python
import os

MODEL = "htdemucs"
TWO_STEMS = False
output_dir = "demucs_output"

extra = "--two-stems vocals" if TWO_STEMS else ""
!demucs -n {MODEL} {extra} "{filename}" --out {output_dir}
```

### 4. Listen to the stems

```python
from IPython.display import Audio, display

base = os.path.splitext(filename)[0]
demucs_path = os.path.join(output_dir, MODEL, base)
stems = ["vocals", "no_vocals"] if TWO_STEMS else ["vocals", "drums", "bass", "other"]

for stem in stems:
    path = os.path.join(demucs_path, f"{stem}.wav")
    if os.path.exists(path):
        print(stem)
        display(Audio(path))
```

### 5. Download all stems as a ZIP

```python
import zipfile

zip_filename = f"/content/{base}_stems.zip"
with zipfile.ZipFile(zip_filename, "w") as zipf:
    for stem in stems:
        path = os.path.join(demucs_path, f"{stem}.wav")
        if os.path.exists(path):
            zipf.write(path, arcname=f"{stem}.wav")

files.download(zip_filename)
```
